```python
import numpy as np
import sympy as sp
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import newton
```

```python
def sigmoidal(t) :
    resultado = 1 / (1 + np.exp(-t))
    return resultado
def euler_inv(t) :
    resultado = 1 - (1/(1*np.exp(t)))
    return resultado
```

```python
list_x = np.linspace(-1 , 4 , 999)
list_fx = [sigmoidal(x) for x in list_x]
list_fx2 = [euler_inv(x) for x in list_x]

plt.figure(figsize=(8,5), dpi=150)
plt.plot(list_x, list_fx, label = "f(x) = 1 / (1 + exp(-x))")
plt.plot(list_x, list_fx2, label = "f2(x) = 1 - 1 / exp(x)")
plt.title("Gráficas de función sigmoidal y función exponencial (euler)")
plt.xlabel("x")
plt.ylabel("f(x) , f2(x)", rotation=90)
plt.xticks(rotation=0)
plt.yticks(np.arange( int(min(list_fx2)), 1.5, 0.5) )
plt.axvline(x=0, color="black", linestyle="--", label="Asintota en el x = 0", linewidth="0.75", alpha=0.5)
plt.axhline(y=1, color="red", linestyle="--", label="Asintota en el y=1", linewidth="0.75", alpha=0.5)
plt.grid(True,linewidth="0.25", alpha=0.4)
plt.annotate("Punto (3, 0.95) aprox.", xy=(3, 0.95), xytext=(2,0), arrowprops = dict(facecolor="black", shrink=0.05))
# plt.legend()
plt.legend(loc="center left", bbox_to_anchor=(1,0.5))
plt.tight_layout()
plt.show()
```
