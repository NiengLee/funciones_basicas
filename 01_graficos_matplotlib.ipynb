```python
import numpy as np
import sympy as sp
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def sigmoidal(t) :
    resultado = 1 / (1 + np.exp(-t))
    return resultado
def euler_inv(t) :
    resultado = 1 - (1/(1*np.exp(t)))
    return resultado
```

```python
list_x = np.linspace(-1, 4, 999)
list_fx = sigmoidal(list_x)
list_f2x = euler_inv(list_x)

plt.figure(figsize=(7, 7), dpi=120)

plt.plot(list_x, list_fx, label="Sigmoidal: 1 / (1 + e^{-x})")
plt.plot(list_x, list_f2x, label="Euler inv: 1 - e^{-x}")

plt.title("f(x) vs x")
plt.xlabel("x", rotation=0)
plt.ylabel("f(x)")
plt.xticks(np.arange(int(min(list_x)), int(max(list_x)) + 1, 1))
plt.yticks(np.arange(int(min(list_f2x))-0.5, int(max(list_fx)) + 1.5, 0.5))

plt.axvline(x=0, color="black", linestyle="--", label="Asintota vertical x = 0", alpha=0.75)
plt.axhline(y=1, color="red", linestyle="--", label="Asintota horizontal y = 1", alpha=0.75)

plt.grid(True, alpha=0.25, linewidth=0.50)
plt.annotate('intersección P(3,0.96) aprox.', xy=(3, 0.96), xytext=(2, 0.5),arrowprops=dict(facecolor='green', shrink=0.05),)
# plt.legend(loc="center left", bbox_to_anchor=(1, 0.5))
plt.legend()
plt.tight_layout()

plt.show()
```
