# 01.Operaciones con derivadas

### Sea $f(x) = x^2 + \frac{1}{x}$, determinar $f'(4)$

```python
print("Realizado por Exergia Code : ")
import sympy as sp
from scipy.integrate import quad
x = sp.Symbol("x")

f = x**2 + 1/x

df = sp.diff(f, x)              
df_simpl = sp.simplify(df)      

val_4 = df_simpl.subs(x, 4)


print(f"f(x) = {f}")
print(f"f'(x) = {df_simpl}")
print(f"f'(4) = {val_4}")
print(f"f'(4) float = {float(val_4)}")
```

# 02.Operaciones con integrales

### Determinar el resultado de la siguiente integral definida :

$$
\int_2^{4} x\mathrm{d}x
$$

```python
print("Realizado por Exergia Code : ")
import sympy as sp
from scipy.integrate import quad
x_i = 2
x_f = 4
x = sp.Symbol('x')
expresion = x

integral_operation = sp.integrate(expresion, (x, x_i, x_f))
print(f"Resultado de la integral entre {x_i} y {x_f} = {integral_operation}")

#   Para integrales complicadas:
res, err = quad(lambda x : x, 2, 4)
print(f"Resultado de la integral entre {x_i} y {x_f} = {res} | error estimado = {err:,.5f}")

# integral indefinida
x = sp.Symbol("x")

F = sp.integrate(x**3, x)   
C = sp.Symbol("C")

print(f"Función resultante integral indefinida : {F}")                 
print(f"Función resultante integral indefinida con constante de integración : {F} + {C}")             
```

# 03.Calculo del valor de una incognita cuando la incognita es un límite de integración

## 1) Determinar un valor "a" tal que:

$$
\int_a^{2} 3x^{2}\mathrm{d}x = 7
$$

## 2) Determinar un valor "b" tal que:

$$
\int_{6}^{b} x\mathrm{d}x = 14
$$

```python
print("Realizado por Exergia Code : ")
from scipy.integrate import quad
from scipy.optimize import newton

# 1)
b_1 = 2
objetivo = 7

def f(x) :
    y = 3 * (x**2)
    return y

def g(c, objetivo) :
    resultado, _ = quad(f,c,b_1)
    return resultado - objetivo

a_1 = newton(g, x0=b_1, args=(objetivo,))
a_1 = round(a_1)
print(f"El limite de integración inferior para el punto 1 es : {a_1}")


# 2)
a_2 = 6
objetivo = 14

def h(x) :
    y = x
    return y

def g(c, objetivo) :
    resultado, _ = quad(h,a_2,c)
    return resultado - objetivo

b_2 = newton(g, x0=10,args=(objetivo,))
b_2 = round(b_2)

print(f"El limite de integración superior para el punto 2 es : {b_2}")
```

# Solucion de ecuaciones no lineales nxn

Uso de fsolve para solución de ecuaciones no lineales, por diferencias finitas para cada paso

```python
def f(x):
  f1 = x[0] + 2*x[1]
  f2 = 1.5*x[0]- x[1] -0.5
  return [f1,f2]

p_0 = [1,1]
x = fsolve(f, p_0)
print(f"Solución del sistema : {x}")
```

```python
def g(x):
    g1 = 2*(x[0]**2) + (x[1]**2) - 1
    g2 = ((0.5*x[0] - 0.5)**2) + 2*((x[1] - 0.25)**2) - 1
    return [g1, g2]

init = [1,1]
roots = fsolve(g, init)
print(f"Solución del sistema : {roots}")
```

```python
def g(z):
    t, x1 = z
    x0 = np.exp(t) - 2.0
    return np.array([
        x0**5 + 2*x0**4 - x1,
        np.log(x0 + 2.0) + 1.0 - x1
    ], dtype=float)

init = [0.0, 2.0]   # t=0 -> x0=-1
roots, info, ier, msg = fsolve(g, init, full_output=True)
t, x1 = roots
x0 = np.exp(t) - 2.0
print("x0, x1:", x0, x1, "ier:", ier, "msg:", msg)
print("residual:", g(roots))
```
