# Métodos de ecuaciones diferenciales

```python
import numpy as np
from scipy.integrate import solve_ivp
```

```python
k = 0.0525

def decay(t, Q):
    return -k * Q

Q0 = [50]  # mg
Q_med = Q0[0]/2

t_int = (0, 10) # años

sol = solve_ivp(decay, t_int, Q0, dense_output=True)

Q_10 = sol.y[0, -1]

# Vida media numérica (buscar t tal que Q=25)
t_vals = np.linspace(0, 50, 10000)
Q_vals = sol.sol(t_vals)[0]
tau_numeric = t_vals[np.argmin(np.abs(Q_vals - Q_med))]

print(f"Cantidad a los 10 años (numérica): {Q_10:.6f}")
print(f"Vida media aproximada (numérica): {tau_numeric:.6f}")
```
